```python
from llama_cpp import Llama

llm = Llama(
    model_path=r"Qwen3.5-9B-UD-Q6_K_XL.gguf",
    n_gpu_layers=-1,
    verbose=False,
    n_ctx=4096,
)
```

```python
role_content = "You are an expert English lexicographer and emotion researcher. Just give one word as the answer."

def classify_emotion(word, rc):
    question = f"Is {word} an positive, negative, neutral, or complex emotion?."
    return llm.create_chat_completion(
        messages=[
            {
                "role": "system",
                "content": rc
            },
            {
                "role": "user",
                "content": question
            }
        ],
        max_tokens=2,
        temperature=0
    )["choices"][0]["message"]["content"].strip().lower()
    
def give_intesity(word, rc):
    
    question = f"""Give intesity score to, {word}. 
                Scale: 0 = not emotional 
                1, 2 = very weak 
                3, 4 = mild 
                5, 6 = moderate 
                7, 8 = strong 
                9, 10 = extremely intense"""
    return llm.create_chat_completion(
        messages=[
            {
                "role": "system",
                "content": rc
            },
            {
                "role": "user",
                "content": question
            }
        ],
        max_tokens=2,
        temperature=0
    )["choices"][0]["message"]["content"].strip().lower()

def further_categorize(word, rc):
    question = f"Give category to the emotion, {word}."
    return llm.create_chat_completion(
        messages=[
            {
                "role": "system",
                "content": rc
            },
            {
                "role": "user",
                "content": question
            }
        ],
        max_tokens=2,
        temperature=0
    )["choices"][0]["message"]["content"].strip().lower()
```

```python
data = []

for word in emotions_words:
    data.append([word, classify_emotion(word, role_content), give_intesity(word, role_content), further_categorize(word, role_content)])
```

```python
import pandas as pd

df = pd.DataFrame(data = data, columns= ["word", "type", "intesity", "category"])
```

```python
import plotly.express as px

counts = df["type"].value_counts().reset_index()
counts.columns = ["type", "count"]

fig = px.bar(
    counts,
    x="type",
    y="count",
    title="Count by Word Type",
    labels={
        "type": "Word Type",
        "count": "Count"
    }
)

fig.show()
```

```python
intensity_counts = df["intesity"].value_counts().sort_index().reset_index()
intensity_counts.columns = ["intesity", "count"]

fig2 = px.bar(
    intensity_counts,
    x="intesity",
    y="count",
    title="Count by Intensity",
    labels={"intesity": "Intensity", "count": "Count"}
)

fig2.show()
```

```python
category_counts = df["category"].value_counts().reset_index()
category_counts.columns = ["category", "count"]

fig3 = px.bar(
    category_counts,
    x="category",
    y="count",
    title="Count by Category",
    labels={"category": "Category", "count": "Count"}
)

fig3.show()
```
